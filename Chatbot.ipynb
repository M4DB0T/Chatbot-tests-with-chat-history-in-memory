```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
groq_api_key=os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(model="llama-3.1-8b-instant",groq_api_key=groq_api_key)
```

```python
from langchain_core.messages import HumanMessage,SystemMessage
model.invoke([HumanMessage(content="Hello My name is Ilkin")])
```

```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables import RunnableWithMessageHistory
```

```python
store={}
def get_session_history(session_id:str)->BaseChatMessageHistory:
    if session_id not in store:
        store[session_id]=ChatMessageHistory()
    return store[session_id]
with_message_history = RunnableWithMessageHistory(model,get_session_history)
```

```python
config={"configurable":{"session_id":"chat1"}}
```

```python
with_message_history.invoke(
    HumanMessage(content="Hello my name is Ilkin and I am senior AI engineer"),
    config=config
)
```

```python
with_message_history.invoke(
    HumanMessage(content="What is my name?"),
    config=config
)
```

```python
from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder

prompt= ChatPromptTemplate.from_messages(
    [
        ("system","you are helpful assistant . answeer all the questions to the best of your ability"),
        MessagesPlaceholder(variable_name="messages")
    ]
)

chain = prompt|model
```

```python
chain.invoke({"messages":[HumanMessage(content="Hello my nmae is Ilkin")]})
```

```python
chain.invoke({"messages":[HumanMessage(content="What is the main causes of cancer")]})
```

```python
with_message_history=RunnableWithMessageHistory(chain,get_session_history)
```

```python
config={"configurable":{"session_id":"chat1"}}
response=with_message_history.invoke(
    [HumanMessage(content="What is the main causes of cancer")],
    config=config

)
```

```python
len(response.content)
```

```python
prompt= ChatPromptTemplate.from_messages(
    [
        (
            "system",
            "you are helpful assistant . answeer all the questions to the best of your ability in {language}"
        ),
        MessagesPlaceholder(variable_name="messages")
    ]
)

chain = prompt|model
```

```python
response=chain.invoke({"messages":[HumanMessage(content="Hi my name is Ilkin")],"language":"Azerbaijani"})
```

```python
response.content
```

```python
with_message_history = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="messages"
)
```

```python
config={"configurable":{"session_id":"chat2"}}
response = with_message_history.invoke(
    {"messages":[HumanMessage(content="Hello My name is Ilkin")],
     "language":"Turkish"},
     config=config
)
response.content
```

```python
from langchain_core.messages import AIMessage,SystemMessage,trim_messages
trimmer=trim_messages(
    max_tokens=45,
    strategy="last",
    token_counter=model,
    include_system=True,
    allow_partial=False,
    start_on="human"
)
messages = [
    SystemMessage(content="you're a good assistant"),
    HumanMessage(content="hi! I'm bob"),
    AIMessage(content="hi!"),
    HumanMessage(content="I like vanilla ice cream"),
    AIMessage(content="nice"),
    HumanMessage(content="whats 2 + 2"),
    AIMessage(content="4"),
    HumanMessage(content="thanks"),
    AIMessage(content="no problem!"),
    HumanMessage(content="having fun?"),
    AIMessage(content="yes!"),
]
trimmer.invoke(messages)
```

```python
from operator import itemgetter
from langchain_core.runnables import RunnablePassthrough


chain=(
    RunnablePassthrough.assign(messages=itemgetter("messages")|trimmer)
    |prompt
    |model
)
chain.invoke({"messages":messages+[HumanMessage(content="what math problem did i asked ?")],"language":"Turkish"})
```

```python
config={"configurable":{"session_id":"chat3"}}
with_message_history=RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="messages",
    config=config

)
response=with_message_history.invoke({
    "messages":messages+[HumanMessage(content="what math problem did i asked ?")],
    "language":"Turkish"
    })
response.content
```
